```python
from itergen.main import IterGen

# Define the grammar
grammar = """
start: paragraph
paragraph: sentence+
sentence: word+ sentence_end
word: /[a-zA-Z0-9]+/ | other_punctuations
sentence_end: "." | "!" | "?"
other_punctuations: "," | ";" | ":" | "'"
%ignore " "
"""

grammar= "python_var_tracking"
model_id = "microsoft/Phi-3-mini-128k-instruct"
model_id = "microsoft/phi-2"



# Initialize IterGen with the grammar and a model with Hugging Face model ID
iter_gen = IterGen(grammar=grammar, model_id=model_id, max_tokens=50, device="cuda", parse_output_only=False)
prompt = "# Write a python script that generates the n'th Fibonacci number.\n"

# Start generation
iter_gen.start(prompt)

# Generate one sentence
generated_sentence = iter_gen.forward()
print("Generated Sentence:", generated_sentence)

# Backtrack by 2 words
current_words = iter_gen.backward("token", num=20)

# Inspect all words in the current generation
# current_words = iter_gen.view("token")
print("Current Words:", current_words)
```
