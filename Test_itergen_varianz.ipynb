```python
from itergen.main import IterGen
from tqdm import tqdm

model = "microsoft/Phi-2"
grammar = """
start: paragraph+
paragraph: sentence+
sentence: word+ sentence_end
word: /[a-zA-Z0-9]+/ | other_punctuations
sentence_end: "." | "!" | "?"
other_punctuations: "," | ";" | ":" | "'" 
%ignore " "
"""

max_tokens = 250
# Initialize IterGen with the grammar and a model with Hugging Face model ID
iter_gen = IterGen(grammar = grammar, model_id=model, max_tokens=max_tokens, recurrence_penalty=0.1)

prompt = "generate me a truly random bedtime story"

# Start generation
iter_gen.start(prompt)

results = []

tries = 10
for i in tqdm(range(tries)):

    generated_sentence = iter_gen.forward()
    iter_gen.backward("token", max_tokens)
    results.append(''.join(generated_sentence))


for result in results:
    print(result)
    
unique_results = len(set(results))
print(f"{unique_results=}")
```
